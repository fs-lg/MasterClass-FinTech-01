```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import sklearn 
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv(r"D:\FinTech\MasterClass-FinTech-01\session_3\data.csv")
df.head()
```

```python
d = {'UK': 0, 'USA': 1, 'N': 2}
df['Nationality'] = df['Nationality'].map(d)
d = {'YES': 1, 'NO': 0}
df['Go'] = df['Go'].map(d)
df
```

```python
features = ['Age', 'Experience', 'Rank', 'Nationality']

X = df[features]
y = df['Go']

display(X)
y
```

```python
dtree = DecisionTreeClassifier()
dtree = dtree.fit(X,y)
tree.plot_tree(dtree, feature_names=features)
```

```python
import pandas as pd

X_new = pd.DataFrame(
    [[35,10,9,1]],
    columns=df.columns[:-1]
)

print(dtree.predict(X_new))
```
